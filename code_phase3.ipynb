```python
# -*- coding: utf-8 -*-

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Bidirectional
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

import joblib
import json


# -----------------------------
# Load data
# -----------------------------
os.listdir("/content")

df = pd.read_csv("/content/salah.csv.csv")
df.head()
df.columns.tolist()

date_col = "Date"
target_col = "Gls"

# 1) تاريخ + ترتيب زمني
df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
df = df.dropna(subset=[date_col]).sort_values(date_col).reset_index(drop=True)

# 2) تنظيف Gls (فيه نصوص مثل: "On matchday squad, but did not play")
df[target_col] = pd.to_numeric(df[target_col], errors="coerce")

# الأفضل: نحذف مباريات ما لعبها
df_clean = df.dropna(subset=[target_col]).reset_index(drop=True)

# 3) استخراج السلسلة y
y = df_clean[target_col].to_numpy(dtype=float).reshape(-1, 1)

print("Rows before cleaning:", len(df))
print("Rows after cleaning :", len(df_clean))
print("First 10 goals:", y[:10].reshape(-1))


# -----------------------------
# Split (70/15/15) - time based
# -----------------------------
n = len(y)
train_end = int(n * 0.70)
val_end   = int(n * 0.85)

y_train = y[:train_end]
y_val   = y[train_end:val_end]
y_test  = y[val_end:]

print("Train/Val/Test sizes:", len(y_train), len(y_val), len(y_test))


# -----------------------------
# Scaling (fit train only)
# -----------------------------
scaler = MinMaxScaler()
y_train_s = scaler.fit_transform(y_train)
y_val_s   = scaler.transform(y_val)
y_test_s  = scaler.transform(y_test)

print("Scaled range (train):", y_train_s.min(), "to", y_train_s.max())


# -----------------------------
# Sequences
# -----------------------------
def make_sequences(series, window):
    X, y_out = [], []
    for i in range(window, len(series)):
        X.append(series[i-window:i])
        y_out.append(series[i])
    return np.array(X), np.array(y_out)

windows = [3, 5, 7]

seq_data = {}
for w in windows:
    X_tr, y_tr = make_sequences(y_train_s, w)
    X_va, y_va = make_sequences(y_val_s, w)
    X_te, y_te = make_sequences(y_test_s, w)

    seq_data[w] = (X_tr, y_tr, X_va, y_va, X_te, y_te)
    print(f"Window={w} -> Train:{X_tr.shape}, Val:{X_va.shape}, Test:{X_te.shape}")


# -----------------------------
# Metrics
# -----------------------------
def eval_metrics(y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae  = mean_absolute_error(y_true, y_pred)
    r2   = r2_score(y_true, y_pred)
    return rmse, mae, r2


# -----------------------------
# (Optional) Naive baseline - OLD (NOT aligned with window)
# Keep it if you want, but DON'T use it in the final comparison.
# -----------------------------
naive_pred_s = y_test_s[:-1]
naive_true_s = y_test_s[1:]

naive_pred = scaler.inverse_transform(naive_pred_s)
naive_true = scaler.inverse_transform(naive_true_s)

rmse_n_old, mae_n_old, r2_n_old = eval_metrics(naive_true, naive_pred)

print("\nNaive Baseline (OLD - not window aligned)")
print("RMSE:", rmse_n_old)
print("MAE :", mae_n_old)
print("R2  :", r2_n_old)


# -----------------------------
# LSTM (w=5)
# -----------------------------
w = 5
X_tr, y_tr, X_va, y_va, X_te, y_te = seq_data[w]

model_lstm = Sequential([
    LSTM(64, input_shape=(w,1)),
    Dense(1)
])

model_lstm.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse"
)

history_lstm = model_lstm.fit(
    X_tr, y_tr,
    validation_data=(X_va, y_va),
    epochs=40,
    batch_size=16,
    verbose=1
)

lstm_pred_s = model_lstm.predict(X_te, verbose=0)
lstm_pred = scaler.inverse_transform(lstm_pred_s)
lstm_true = scaler.inverse_transform(y_te)

rmse_l, mae_l, r2_l = eval_metrics(lstm_true, lstm_pred)

print("\nLSTM Results (w=5)")
print("RMSE:", rmse_l)
print("MAE :", mae_l)
print("R2  :", r2_l)


# -----------------------------
# BiLSTM (w=5)
# -----------------------------
early = EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True)

model_bilstm = Sequential([
    Bidirectional(LSTM(64), input_shape=(w,1)),
    Dense(1)
])

model_bilstm.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse"
)

history_bilstm = model_bilstm.fit(
    X_tr, y_tr,
    validation_data=(X_va, y_va),
    epochs=60,
    batch_size=16,
    callbacks=[early],
    verbose=1
)

bilstm_pred_s = model_bilstm.predict(X_te, verbose=0)
bilstm_pred = scaler.inverse_transform(bilstm_pred_s)
bilstm_true = scaler.inverse_transform(y_te)

rmse_b, mae_b, r2_b = eval_metrics(bilstm_true, bilstm_pred)

print("\nBiLSTM Results (w=5)")
print("RMSE:", rmse_b)
print("MAE :", mae_b)
print("R2  :", r2_b)


# -----------------------------
# Hyperparameter tuning (as in your code)
# -----------------------------
def train_lstm(X_tr, y_tr, X_va, y_va, units=32, lr=1e-3, epochs=40):
    model = Sequential([
        LSTM(units, input_shape=(X_tr.shape[1], 1)),
        Dense(1)
    ])
    model.compile(optimizer=Adam(learning_rate=lr), loss="mse")
    early_tune = EarlyStopping(monitor="val_loss", patience=6, restore_best_weights=True)
    model.fit(
        X_tr, y_tr,
        validation_data=(X_va, y_va),
        epochs=epochs,
        batch_size=16,
        callbacks=[early_tune],
        verbose=0
    )
    return model

tuning_results = []

for w_ in [3,5,7]:
    X_tr_, y_tr_, X_va_, y_va_, X_te_, y_te_ = seq_data[w_]
    for units in [32, 64]:
        for lr in [1e-3, 5e-4]:
            m = train_lstm(X_tr_, y_tr_, X_va_, y_va_, units=units, lr=lr, epochs=50)
            val_pred_s = m.predict(X_va_, verbose=0)
            val_pred = scaler.inverse_transform(val_pred_s)
            val_true = scaler.inverse_transform(y_va_)
            rmse_v, mae_v, r2_v = eval_metrics(val_true, val_pred)
            tuning_results.append((w_, units, lr, rmse_v, mae_v, r2_v))
            print(f"w={w_}, units={units}, lr={lr} -> Val RMSE={rmse_v:.4f}, MAE={mae_v:.4f}, R2={r2_v:.4f}")

best = sorted(tuning_results, key=lambda x: x[3])[0]
print("\nBest tuning (by Val RMSE):", best)


# ==========================================================
# ✅ FINAL COMPARISON (ALIGNED NAIVE + BETTER PLOTS)
# ==========================================================
w = 5
_, _, _, _, X_te, y_te = seq_data[w]

true_test = scaler.inverse_transform(y_te)

# ✅ aligned naive = last value in window
naive_pred_s_w = X_te[:, -1, 0].reshape(-1,1)
naive_pred_w = scaler.inverse_transform(naive_pred_s_w)

# Predictions (already trained)
lstm_pred_s = model_lstm.predict(X_te, verbose=0)
lstm_pred = scaler.inverse_transform(lstm_pred_s)

bilstm_pred_s = model_bilstm.predict(X_te, verbose=0)
bilstm_pred = scaler.inverse_transform(bilstm_pred_s)

# ✅ aligned metrics (USE THESE IN REPORT)
rmse_n, mae_n, r2_n = eval_metrics(true_test, naive_pred_w)
rmse_l2, mae_l2, r2_l2 = eval_metrics(true_test, lstm_pred)
rmse_b2, mae_b2, r2_b2 = eval_metrics(true_test, bilstm_pred)

print("\n--- FINAL (Aligned on w=5 test) ---")
print("Naive  :", (rmse_n, mae_n, r2_n))
print("LSTM   :", (rmse_l2, mae_l2, r2_l2))
print("BiLSTM :", (rmse_b2, mae_b2, r2_b2))

# ✅ Plot 1 + Plot 2 (no contradiction)
err_naive = np.abs(true_test - naive_pred_w)
err_lstm  = np.abs(true_test - lstm_pred)
err_bi    = np.abs(true_test - bilstm_pred)

fig, axs = plt.subplots(2, 1, figsize=(12,7), sharex=True)

# (A) Actual vs Predictions
axs[0].plot(true_test, label="Actual", linewidth=2, marker='o')
axs[0].plot(naive_pred_w, label="Naive", linestyle='--')
axs[0].plot(lstm_pred, label="LSTM")
axs[0].plot(bilstm_pred, label="BiLSTM")
axs[0].set_title("One-Step Prediction (Test Set) - w=5")
axs[0].legend()

# (B) Absolute Error
axs[1].plot(err_naive, label="Naive |Error|")
axs[1].plot(err_lstm,  label="LSTM |Error|")
axs[1].plot(err_bi,    label="BiLSTM |Error|")
axs[1].set_title("Absolute Prediction Error (Lower is Better)")
axs[1].set_xlabel("Time Step")
axs[1].legend()

plt.tight_layout()
plt.show()

# Residual plots (keep yours)
plt.figure(figsize=(12,3))
plt.plot(true_test - lstm_pred, label="Residuals (LSTM)")
plt.title("Residuals Plot: Actual - Predicted (LSTM)")
plt.legend()
plt.show()

plt.figure(figsize=(12,3))
plt.plot(true_test - bilstm_pred, label="Residuals (BiLSTM)")
plt.title("Residuals Plot: Actual - Predicted (BiLSTM)")
plt.legend()
plt.show()


# -----------------------------
# Save outputs
# -----------------------------
os.makedirs("phase3_outputs", exist_ok=True)

model_lstm.save("phase3_outputs/lstm_w5.keras")
model_bilstm.save("phase3_outputs/bilstm_w5.keras")
joblib.dump(scaler, "phase3_outputs/minmax_scaler.joblib")

summary = {
    "dataset": "salah_timeseries.csv.csv",
    "rows_before_cleaning": 291,
    "rows_after_cleaning": int(len(df_clean)),
    "split": {"train": int(len(y_train)), "val": int(len(y_val)), "test": int(len(y_test))},

    # OLD (keep for reference, but not for report)
    "baseline_naive_old_not_aligned": {"rmse": float(rmse_n_old), "mae": float(mae_n_old), "r2": float(r2_n_old)},

    # ✅ FINAL aligned metrics (USE THESE)
    "baseline_naive_aligned_w5": {"rmse": float(rmse_n), "mae": float(mae_n), "r2": float(r2_n)},
    "lstm_w5_aligned": {"rmse": float(rmse_l2), "mae": float(mae_l2), "r2": float(r2_l2)},
    "bilstm_w5_aligned": {"rmse": float(rmse_b2), "mae": float(mae_b2), "r2": float(r2_b2)},

    "tuning_results": [{"window": w, "units": u, "lr": lr, "val_rmse": float(rm), "val_mae": float(ma), "val_r2": float(r2v)}
                       for (w,u,lr,rm,ma,r2v) in tuning_results]
}

with open("phase3_outputs/phase3_summary.json", "w") as f:
    json.dump(summary, f, indent=2)

print("\nSaved outputs in phase3_outputs/")
print(os.listdir("phase3_outputs"))
```
